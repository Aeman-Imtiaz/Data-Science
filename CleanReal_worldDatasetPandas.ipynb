**errors="coerce"**

errors="coerce" is used in Pandas when we want to convert values to a particular data type but some values may be invalid.

For example:

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
What does it do?

Suppose the CSV contains:

Age
20
21
twenty-two
19

Normally, "twenty-two" cannot be converted to a number.

With:

errors="coerce"

Pandas converts invalid values into NaN:

Age
20
21
NaN
19

Then we can handle the missing value:

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

df["Age"] = df["Age"].fillna(df["Age"].median())

**Three important options**

Option	What happens

**errors="raise"**	Stops and gives an error

**errors="coerce"**	Invalid value → NaN

**errors="ignore"**	Leaves the original value unchanged


errors="coerce" tells Pandas: "If a value cannot be converted, don't stop the program; replace it with NaN."

**This is particularly useful when cleaning real-world CSV datasets, because real data often contains invalid or unexpected values.**

**Numeric conversion is required to ensure that numerical data is stored as numbers rather than text, allowing mathematical calculations and machine-learning algorithms to work correctly.**

```python
# 6. Convert numeric columns
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Study_Hours"] = pd.to_numeric(df["Study_Hours"], errors="coerce")
df["Attendance"] = pd.to_numeric(df["Attendance"], errors="coerce")
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")
```

```python
# 7. Fill missing numeric values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Study_Hours"] = df["Study_Hours"].fillna(df["Study_Hours"].median())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
```

```python
# 8. Fix categorical values
df["Gender"] = df["Gender"].replace({
    "M": "Male",
    "male": "Male",
    "F": "Female",
    "female": "Female"
})

df["Department"] = df["Department"].replace({
    "Cyber-Security": "Cyber Security"
})

print(df)
```

```python
#Task 1 — Load and Inspect the Data
from google.colab import files
import pandas as pd

# Pick CSV file from your laptop
uploaded = files.upload()

# Read CSV
df = pd.read_csv("customer_sales.csv")
print(df.head())
print(df.info())
print(df.shape)
```

**Question: Which columns contain missing values?**

```python
#Task 2 — Find Missing Values
print(df.isnull().sum())
```

```python
#Task 3 — Remove Duplicate Orders
df = df.drop_duplicates()
```

```python
#Task 4 — Convert Numeric Columns

#The Quantity and Price columns may contain invalid text values.

df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

#Then check:

print(df.dtypes)
```

```python
#Task 5 — Handle Missing Values

#Replace missing numerical values with the median:

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Price"] = df["Price"].fillna(df["Price"].median())

#For missing city/payment values:

df["City"] = df["City"].fillna("Unknown")
df["Payment"] = df["Payment"].fillna("Unknown")
```

```python
#Task 6 — Standardize Text Data

#Make city and product names consistent.

#For example: .title(): This method capitalizes the first character of each word and lowers all other characters

df["City"] = df["City"].str.title()
df["Product"] = df["Product"].str.title()
df["Payment"] = df["Payment"].str.title()

#This changes:

#rawalpindi → Rawalpindi
#ISLAMABAD  → Islamabad
#mouse      → Mouse
#cash       → Cash
```

```python
#Task 7 — Convert Order Date
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)

#Question: What happened to the invalid date?
```

```python
#Task 8 — Create a New Column

#Calculate the Total Sales for each order:

df["Total_Sales"] = df["Quantity"] * df["Price"]

```

```python
#Task 9 — Find the Best-Selling Product
product_sales = df.groupby("Product")["Total_Sales"].sum()
#.sort_values(ascending=False)

print(product_sales)

#Then find the highest:

print(product_sales.idxmax())
print(product_sales.max())



#Question: Which product generated the highest sales?
```
